# Réponses aux questions sur l'apprentissage profond

## 1. Quelle est la principale différence entre l'apprentissage automatique traditionnel et l'apprentissage profond ?

La principale différence réside dans la manière dont les caractéristiques (features) sont obtenues.

- **Apprentissage automatique traditionnel (Machine Learning)** :
  - Nécessite souvent une sélection et une création manuelles des caractéristiques pertinentes (ingénierie des caractéristiques).
  - Utilise des algorithmes comme les arbres de décision, les SVM ou les forêts aléatoires.

- **Apprentissage profond (Deep Learning)** :
  - Utilise des réseaux de neurones à plusieurs couches pour apprendre automatiquement les caractéristiques à partir des données brutes.
  - Est particulièrement efficace pour les données complexes comme les images, les textes et les sons.

**En résumé :** le Machine Learning nécessite généralement une intervention humaine pour extraire les caractéristiques, tandis que le Deep Learning les apprend automatiquement.

---

## 2. Comment les réseaux de neurones artificiels (RNA) imitent-ils le cerveau humain ?

Les RNA s'inspirent du fonctionnement simplifié des neurones biologiques.

- Chaque neurone artificiel reçoit des informations en entrée.
- Il les pondère grâce à des poids.
- Il applique une fonction d'activation pour produire une sortie.
- Les neurones sont interconnectés en couches :
  - couche d'entrée ;
  - couches cachées ;
  - couche de sortie.

Pendant l'apprentissage, les poids sont ajustés pour améliorer les prédictions, de manière analogue au renforcement des connexions entre neurones dans le cerveau.

---

## 3. Pourquoi l'apprentissage profond est-il plus performant sur les grands ensembles de données que l'apprentissage automatique traditionnel ?

L'apprentissage profond est plus performant car :

- Il peut apprendre des représentations très complexes des données.
- Il extrait automatiquement des caractéristiques de haut niveau.
- Ses performances augmentent généralement avec la quantité de données disponibles.
- Il exploite efficacement la puissance des processeurs graphiques (GPU).

Les modèles traditionnels atteignent souvent un plateau de performance lorsque les données deviennent très volumineuses, contrairement aux modèles profonds.

---

## 4. Quels sont les défis de l'apprentissage profond, et comment peut-on les relever ?

### Défis

1. Besoin de grandes quantités de données.
2. Temps d'entraînement élevé.
3. Forte consommation de ressources informatiques.
4. Risque de surapprentissage (overfitting).
5. Difficulté d'interprétation des modèles.

### Solutions

- Utiliser l'augmentation de données (Data Augmentation).
- Employer des GPU ou TPU pour accélérer les calculs.
- Appliquer des techniques de régularisation comme le Dropout.
- Utiliser le transfert d'apprentissage (Transfer Learning).
- Développer des méthodes d'IA explicable (Explainable AI).

---

## 5. Qu’est-ce que l’ingénierie des caractéristiques, et pourquoi n’est-elle pas nécessaire en apprentissage profond ?

L'ingénierie des caractéristiques consiste à sélectionner, transformer et créer des variables pertinentes à partir des données brutes afin d'améliorer les performances d'un modèle.

En Machine Learning traditionnel :
- cette étape est souvent indispensable ;
- elle dépend fortement de l'expertise humaine.

En Deep Learning :
- les réseaux apprennent automatiquement les caractéristiques importantes grâce à leurs multiples couches cachées ;
- cela réduit considérablement le besoin d'intervention humaine.

Ainsi, le Deep Learning automatise une grande partie de l'ingénierie des caractéristiques.

---

## 6. Quel rôle jouent les couches cachées dans un modèle d'apprentissage profond ?

Les couches cachées permettent d'apprendre des représentations de plus en plus abstraites des données.

Par exemple, dans la reconnaissance d'images :

- première couche : détecte les contours ;
- deuxième couche : détecte les formes ;
- troisième couche : détecte les objets ;
- couches profondes : reconnaissent des objets complexes comme des visages ou des véhicules.

Plus il y a de couches cachées, plus le réseau peut modéliser des relations complexes.

Leur rôle principal est donc d'extraire automatiquement les caractéristiques utiles pour effectuer des prédictions précises.

---

## 7. Dans un réseau neuronal artificiel (RNA), quelle est la fonction d'une fonction d'activation ?

Une fonction d'activation détermine si un neurone doit être activé ou non en transformant la somme pondérée des entrées.

### Ses principaux rôles sont :

- Introduire de la non-linéarité dans le réseau.
- Permettre l'apprentissage de relations complexes.
- Contrôler la sortie des neurones.
- Faciliter l'apprentissage grâce à des propriétés mathématiques adaptées.

### Exemples courants :

| Fonction | Utilisation |
|----------|------------|
| Sigmoïde | Classification binaire |
| Tanh | Sorties comprises entre -1 et 1 |
| ReLU | Très utilisée dans les couches cachées |
| Leaky ReLU | Réduit le problème des neurones morts |
| Softmax | Classification multiclasse |

Sans fonction d'activation, un réseau neuronal ne serait qu'un modèle linéaire, incapable d'apprendre des relations complexes entre les données.

---

# Conclusion

L'apprentissage profond est une évolution du Machine Learning qui utilise des réseaux de neurones multicouches pour apprendre automatiquement les caractéristiques des données. Grâce aux couches cachées et aux fonctions d'activation, il peut résoudre des problèmes complexes avec une grande précision, notamment dans les domaines de la vision par ordinateur, du traitement du langage naturel et de la reconnaissance vocale.

# Exercice : Système de décision basé sur un perceptron

## Données

- **Poids de la température** : \(w_1 = 0,6\)
- **Poids de la pluie** : \(w_2 = 0,4\)
- **Biais** : \(b = 2\)

### Formule de la somme pondérée

\[
S = (\text{Température} \times 0,6) + (\text{Pluie} \times 0,4) + 2
\]

### Fonction d'activation

- Si \(S > 20\), alors **sortie = 1** (Oui, sortir).
- Si \(S \leq 20\), alors **sortie = 0** (Non, rester à l'intérieur).

---

# 1. Calcul de la sortie

## Cas 1 : Température = 70°F, Pluie = 0

### Étape 1 : Calcul de la somme pondérée

\[
S = (70 \times 0,6) + (0 \times 0,4) + 2
\]

\[
S = 42 + 0 + 2
\]

\[
S = 44
\]

### Étape 2 : Application de la fonction d'activation

Comme :

\[
44 > 20
\]

La sortie est :

\[
\boxed{1}
\]

**Décision : Oui, sortir.**

---

## Cas 2 : Température = 50°F, Pluie = 1

### Étape 1 : Calcul de la somme pondérée

\[
S = (50 \times 0,6) + (1 \times 0,4) + 2
\]

\[
S = 30 + 0,4 + 2
\]

\[
S = 32,4
\]

### Étape 2 : Application de la fonction d'activation

Comme :

\[
32,4 > 20
\]

La sortie est :

\[
\boxed{1}
\]

**Décision : Oui, sortir.**

---

# 2. Interprétation des résultats

| Cas | Température | Pluie | Somme pondérée | Sortie | Décision |
|-----|------------|--------|----------------|---------|----------|
| 1 | 70°F | Non (0) | 44 | 1 | Sortir |
| 2 | 50°F | Oui (1) | 32,4 | 1 | Sortir |

## Analyse

Le perceptron suggère de **sortir dans les deux cas**.

### Pourquoi ?

- Dans le premier cas, la température élevée (70°F) contribue fortement à la somme pondérée, qui dépasse largement le seuil de 20.
- Dans le second cas, malgré la pluie, la température de 50°F reste suffisamment élevée et, avec le biais, la somme pondérée dépasse également le seuil.

Le facteur **température**, ayant un poids important (0,6) et des valeurs numériques élevées, influence beaucoup plus la décision que la variable **pluie**, dont la contribution maximale n'est que de 0,4.

## Conclusion

Le perceptron recommande de sortir dans les deux situations, car les sommes pondérées obtenues (44 et 32,4) sont toutes deux supérieures au seuil d'activation de 20. Dans ce modèle simplifié, la température a un impact beaucoup plus important sur la décision que la présence de pluie.

## Exercice 3

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
print(x_train.shape)
print(y_train.shape)

print(x_test.shape)
print(y_test.shape)
```

```python
x_train = x_train / 255.0
x_test = x_test / 255.0
```

```python
num_classes = 10
y_train = to_categorical(y_train, num_classes)
y_test = to_categorical(y_test, num_classes)
```

```python
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)

print("Perte :", test_loss)
print("Précision :", test_accuracy)
```

# Exercice : Propagation avant d'un réseau neuronal simple

## Données

### Entrées
- **x₁ = 2000** (Superficie en pieds carrés)
- **x₂ = 3** (Nombre de chambres)

### Poids
- **w₁ = 0.5**
- **w₂ = 0.7**

### Biais
- **b = 50 000**

### Fonction d'activation
- **ReLU** : `ReLU(z) = max(0, z)`

---

## 1. Calcul de la valeur de sortie avant activation (z)

La formule de la propagation avant est :

\[
z = w_1x_1 + w_2x_2 + b
\]

En remplaçant les valeurs :

\[
z = (0.5 \times 2000) + (0.7 \times 3) + 50000
\]

Calcul des contributions :

- `0.5 × 2000 = 1000`
- `0.7 × 3 = 2.1`

Donc :

\[
z = 1000 + 2.1 + 50000
\]

\[
z = 51002.1
\]

**Valeur avant activation :**

\[
\boxed{z = 51002.1}
\]

---

## 2. Application de la fonction ReLU

La fonction ReLU est définie par :

\[
ReLU(z) = \max(0, z)
\]

Comme :

\[
51002.1 > 0
\]

Alors :

\[
ReLU(51002.1) = 51002.1
\]

**Prédiction finale :**

\[
\boxed{51002.1}
\]

---

## 3. Interprétation

Le réseau neuronal prédit un prix de maison de :

\[
\boxed{51\,002.1}
\]

Si les valeurs sont exprimées en dollars, le prix estimé est de **51 002,10 $**.

---

# Résumé

| Étape | Calcul | Résultat |
|--------|---------|-----------|
| Contribution de la superficie | `0.5 × 2000` | `1000` |
| Contribution des chambres | `0.7 × 3` | `2.1` |
| Ajout du biais | `1000 + 2.1 + 50000` | `51002.1` |
| Application de ReLU | `max(0, 51002.1)` | `51002.1` |

## Réponses finales

### 1. Valeur avant activation
**z = 51002.1**

### 2. Valeur après activation ReLU
**ReLU(z) = 51002.1**

### 3. Prix prédit de la maison
**51 002,10 unités monétaires (par exemple, dollars).**

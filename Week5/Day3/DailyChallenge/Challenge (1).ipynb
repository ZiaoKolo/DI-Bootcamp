### 1. Load the MNIST Dataset

First, we'll load the MNIST dataset using Keras and print the shapes of the loaded data to understand its structure.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.utils import to_categorical

# Load the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Print the shapes of the loaded data
print(f"x_train shape: {x_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"x_test shape: {x_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

### 2. Preprocessing Data for a Fully Connected Network

For a fully connected network, we need to flatten the 28x28 images into a 784-pixel vector, normalize the pixel values to a range of 0-1, and one-hot encode the target labels.

```python
# Flatten images from 28x28 to 784 pixels
x_train_fc = x_train.reshape(-1, 784).astype('float32')
x_test_fc = x_test.reshape(-1, 784).astype('float32')

# Normalize pixel values to [0, 1]
x_train_fc /= 255
x_test_fc /= 255

# One-hot encode target labels
y_train_fc = to_categorical(y_train, 10)
y_test_fc = to_categorical(y_test, 10)

print(f"x_train_fc shape after preprocessing: {x_train_fc.shape}")
print(f"y_train_fc shape after preprocessing: {y_train_fc.shape}")
```

### 3. Build and Train a Fully Connected Network

Now, we'll construct a simple fully connected neural network (also known as a Multilayer Perceptron or MLP) using `Sequential` model with `Dense` layers. We'll compile it with the Adam optimizer and categorical cross-entropy loss, then train it.

```python
# Build the Fully Connected Network model
model_fc = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax') # 10 classes for digits 0-9
])

# Compile the model
model_fc.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display model summary
model_fc.summary()

# Train the model
print("\nTraining Fully Connected Network...")
history_fc = model_fc.fit(
    x_train_fc, y_train_fc,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

# Evaluate the model on the test set
loss_fc, accuracy_fc = model_fc.evaluate(x_test_fc, y_test_fc, verbose=0)
print(f"\nFully Connected Network - Test Accuracy: {accuracy_fc:.4f}")
```

### 4. Preprocessing Data for a Convolutional Neural Network

For a CNN, the input images need to be reshaped to include a channel dimension (e.g., `(batch_size, height, width, channels)`). For grayscale images like MNIST, this will be `(60000, 28, 28, 1)`. We will also normalize the pixel values and one-hot encode the labels again.

```python
# Reshape data to be (batch_size, height, width, channels)
x_train_cnn = x_train.reshape(-1, 28, 28, 1).astype('float32')
x_test_cnn = x_test.reshape(-1, 28, 28, 1).astype('float32')

# Normalize pixel values to [0, 1]
x_train_cnn /= 255
x_test_cnn /= 255

# One-hot encode target labels
y_train_cnn = to_categorical(y_train, 10)
y_test_cnn = to_categorical(y_test, 10)

print(f"x_train_cnn shape after preprocessing: {x_train_cnn.shape}")
print(f"y_train_cnn shape after preprocessing: {y_train_cnn.shape}")
```

### 5. Design and Train a Convolutional Neural Network

Now, let's build a Convolutional Neural Network (CNN) model. This will include `Conv2D` layers for feature extraction, `MaxPooling2D` layers for dimensionality reduction, a `Flatten` layer to transition to dense layers, and finally, `Dense` layers for classification. We'll compile and train it similarly to the fully connected network.

```python
# Build the CNN model
model_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax') # 10 classes for digits 0-9
])

# Compile the model
model_cnn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Display model summary
model_cnn.summary()

# Train the model
print("\nTraining Convolutional Neural Network...")
history_cnn = model_cnn.fit(
    x_train_cnn, y_train_cnn,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

# Evaluate the model on the test set
loss_cnn, accuracy_cnn = model_cnn.evaluate(x_test_cnn, y_test_cnn, verbose=0)
print(f"\nConvolutional Neural Network - Test Accuracy: {accuracy_cnn:.4f}")
```

### 6. Compare the Performance

Finally, let's compare the test accuracies of both the Fully Connected Network and the Convolutional Neural Network.

```python
print("\n--- Model Performance Comparison ---")
print(f"Fully Connected Network Test Accuracy: {accuracy_fc:.4f}")
print(f"Convolutional Neural Network Test Accuracy: {accuracy_cnn:.4f}")

if accuracy_cnn > accuracy_fc:
    print("The Convolutional Neural Network performed better on the MNIST dataset.")
elif accuracy_fc > accuracy_cnn:
    print("The Fully Connected Network performed better on the MNIST dataset.")
else:
    print("Both networks achieved similar performance on the MNIST dataset.")
```

## Exercice 1

# Comparaison entre l'apprentissage automatique traditionnel et l'apprentissage profond

## 1. Tableau comparatif

| Aspect | Apprentissage automatique traditionnel | Apprentissage profond |
|--------|----------------------------------------|----------------------|
| **Ingénierie des fonctionnalités** | Nécessite une sélection et une extraction manuelles des caractéristiques. | Extrait automatiquement les caractéristiques importantes à partir des données. |
| **Informatique** | Utilise des algorithmes comme les arbres de décision, SVM et régression. | Utilise des réseaux neuronaux profonds comportant plusieurs couches cachées. |
| **Évolutivité** | Adapté aux petits et moyens ensembles de données. | Performant avec de très grands ensembles de données. |
| **Découverte de modèles** | Identifie des relations simples ou modérément complexes. | Capable de découvrir des relations très complexes et non linéaires. |
| **Exigences de calcul** | Faibles à modérées ; peut fonctionner sur des ordinateurs classiques. | Très élevées ; nécessite souvent des GPU ou TPU pour un entraînement efficace. |

---

## 2. Exemples d'applications

### Cas où l'apprentissage automatique traditionnel est plus adapté

**Détection des courriels indésirables (spam).**

Les données sont souvent structurées et les modèles comme les arbres de décision, Naive Bayes ou les SVM offrent de bonnes performances avec un coût de calcul relativement faible.

### Cas où l'apprentissage profond est préférable

**Reconnaissance d'images médicales (radiographies, IRM).**

Les réseaux neuronaux convolutifs (CNN) peuvent apprendre automatiquement des caractéristiques complexes dans les images et atteindre une précision très élevée.

---

## 3. Pourquoi l'apprentissage profond est-il avantageux pour les données non structurées ?

L'apprentissage profond est particulièrement efficace pour les données non structurées telles que les images, les vidéos, les textes et les fichiers audio. Contrairement aux méthodes traditionnelles, il apprend automatiquement les caractéristiques pertinentes sans nécessiter une extraction manuelle. Grâce à ses multiples couches de neurones, il peut identifier des motifs complexes et des relations cachées dans les données. Cette capacité lui permet d'obtenir d'excellentes performances dans des domaines comme la vision par ordinateur, le traitement du langage naturel et la reconnaissance vocale.

## Exercice 2

# Réseau neuronal artificiel (RNA)

## 1. Schéma d'un réseau neuronal simple

```text
                    Couche cachée
                 (4 neurones, ReLU)
               H1 ●───────────────┐
             ╱  │  ╲             │
            ╱   │   ╲            │
           ╱    │    ╲           │
Entrée    ●─────●─────●         ● O1
 X1       │╲   ╱│╲   ╱│╲       ╱
          │ ╲ ╱ │ ╲ ╱ │ ╲     ╱
Entrée    ●──●──●──●──●──●───● O2
 X2       │ ╱ ╲ │ ╱ ╲ │ ╱
          │╱   ╲│╱   ╲│╱
Entrée    ●─────●─────●
 X3       H2    H3    H4

        Couche d'entrée     Couche de sortie
         (3 neurones)         (2 neurones)
```

---

## 2. Composants du réseau

### a) Neurones (Nodes)

- Couche d'entrée : **X1, X2, X3**
- Couche cachée : **H1, H2, H3, H4**
- Couche de sortie : **O1, O2**

Les neurones reçoivent des informations, effectuent des calculs et transmettent les résultats à la couche suivante.

---

### b) Poids (Weights)

Les connexions entre deux neurones sont appelées des poids.

Exemples :

- W1 : X1 → H1
- W2 : X2 → H1
- W3 : H1 → O1

Les poids déterminent l'importance de chaque entrée.

---

### c) Biais (Bias)

Chaque neurone possède un biais, généralement noté **b**.

Il permet de déplacer la fonction d'activation et d'améliorer la capacité d'apprentissage du réseau.

Calcul d'un neurone :

\[
z = \sum (w_i x_i) + b
\]

---

### d) Fonction d'activation

Après le calcul de z, une fonction d'activation est appliquée.

Exemples :

- ReLU :

\[
f(x)=\max(0,x)
\]

- Sigmoid :

\[
f(x)=\frac{1}{1+e^{-x}}
\]

Dans ce réseau :
- Couche cachée : ReLU.
- Couche de sortie : Sigmoid (classification binaire) ou Softmax (classification multiple).

---

### e) Couches

Le réseau comporte trois couches :

| Couche | Nombre de neurones |
|---------|-------------------|
| Entrée | 3 |
| Cachée | 4 |
| Sortie | 2 |

---

# 3. Circulation de l'information

L'information circule d'abord de la couche d'entrée vers la couche cachée, où chaque neurone effectue une combinaison pondérée des entrées et ajoute un biais. Une fonction d'activation est ensuite appliquée pour introduire de la non-linéarité dans le modèle. Les résultats obtenus sont transmis à la couche de sortie, qui produit les prédictions finales. Pendant l'entraînement, les poids et les biais sont ajustés grâce à la rétropropagation afin de réduire les erreurs et d'améliorer les performances du réseau.

````python
import numpy as np
import matplotlib.pyplot as plt

# 1. Génération des données
np.random.seed(0)
x = np.arange(-1, 1, 0.1)
y = -x**2 + np.random.normal(0, 0.05, len(x))

# 2. Division des données
x_train = x[:12]
y_train = y[:12]

x_test = x[12:]
y_test = y[12:]

# 3. Fonction d'ajustement polynomial
def polynomial_fit(degree):
    """
    Ajuste un polynôme de degré donné
    sur les données d'entraînement.
    """
    coefficients = np.polyfit(x_train, y_train, degree)
    polynomial = np.poly1d(coefficients)
    return polynomial

# 4. Fonction d'affichage
def plot_polyfit(degree):

    # Ajustement du modèle
    model = polynomial_fit(degree)

    # Points pour tracer la courbe
    x_curve = np.linspace(-1, 1, 200)
    y_curve = model(x_curve)

    # Création de la figure
    plt.figure(figsize=(8,5))

    # Données d'entraînement
    plt.scatter(
        x_train,
        y_train,
        color='blue',
        label='Entraînement'
    )

    # Données de test
    plt.scatter(
        x_test,
        y_test,
        color='red',
        label='Test'
    )

    # Courbe ajustée
    plt.plot(
        x_curve,
        y_curve,
        color='green',
        linewidth=2,
        label=f'Polynôme degré {degree}'
    )

    plt.title(f'Ajustement polynomial (degré {degree})')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.grid(True)

    plt.show()

# 5. Visualisation

plot_polyfit(1)

plot_polyfit(7)

plot_polyfit(11)
````

## Exercice 5

````python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# ==========================
# 1. Génération des données
# ==========================

np.random.seed(42)

X = np.linspace(-3, 3, 100).reshape(-1, 1)
y = -X**2 + np.random.normal(0, 1, X.shape)

# Séparation entraînement / test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# ==========================
# 2. Parcourir les degrés 1 à 11
# ==========================

resultats = []

for degre in range(1, 12):

    # Création du modèle polynomial
    modele = make_pipeline(
        PolynomialFeatures(degre),
        LinearRegression()
    )

    # Entraînement
    modele.fit(X_train, y_train)

    # Prédictions
    y_train_pred = modele.predict(X_train)
    y_test_pred = modele.predict(X_test)

    # Calcul RMSE
    rmse_train = np.sqrt(mean_squared_error(y_train, y_train_pred))
    rmse_test = np.sqrt(mean_squared_error(y_test, y_test_pred))

    # Stockage des résultats
    resultats.append((degre, rmse_train, rmse_test))

# ==========================
# 3. Affichage des résultats
# ==========================

print("Degré | RMSE Train | RMSE Test")
print("-" * 35)

for degre, rmse_train, rmse_test in resultats:
    print(f"{degre:^6} | {rmse_train:.4f} | {rmse_test:.4f}")

# ==========================
# 4. Tracé des courbes RMSE
# ==========================

degres = [r[0] for r in resultats]
rmse_train = [r[1] for r in resultats]
rmse_test = [r[2] for r in resultats]

plt.figure(figsize=(8, 5))
plt.plot(degres, rmse_train, marker='o', label='RMSE Entraînement')
plt.plot(degres, rmse_test, marker='s', label='RMSE Test')

plt.yscale('log')  # échelle logarithmique
plt.xlabel("Degré du polynôme")
plt.ylabel("RMSE (échelle logarithmique)")
plt.title("Évolution de la RMSE selon le degré du polynôme")
plt.legend()
plt.grid(True)
plt.show()

# ==========================
# 5. Meilleur degré
# ==========================

meilleur = min(resultats, key=lambda x: x[2])

print("\nMeilleur degré :", meilleur[0])
print("RMSE Test minimale :", round(meilleur[2], 4))
````

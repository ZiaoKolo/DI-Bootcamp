```python
import numpy as np

#Matrice 3x3
matrice = np.array([[1, 2, 3], [0, 1, 4], [5, 6, 0]])
determinant = np.linalg.det(matrice)
print("Déterminant de la matrice : ", determinant)
inverse = np.linalg.inv(matrice)
print("Inverse de la matrice : \n", inverse)
```

## Exercice 6

```python
import numpy as np
from scipy import stats

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
test_statistic, p_value = stats.ttest_rel(
    productivity_after,
    productivity_before
)

print("Statistique t :", test_statistic)
print("p-value :", p_value)
```

## Exercice 8

```python
import numpy as np

# Génération des dates de l'année 2023
dates = np.arange('2023-01', '2024-01', dtype='datetime64[M]')

# Génération de données aléatoires (par exemple des ventes ou températures)
valeurs = np.random.randint(10, 100, size=12)

print("Dates :")
print(dates)

print("Valeurs :")
print(valeurs)

# Janvier à Mars
jan_mar = valeurs[0:3]
print("Janvier à Mars :")
print(jan_mar)

# Avril à Juin
avr_jun = valeurs[3:6]
print("Avril à Juin :")
print(avr_jun)

# Juillet à Septembre
jul_sep = valeurs[6:9]
print("Juillet à Septembre :")
print(jul_sep)

# Octobre à Décembre
oct_dec = valeurs[9:12]
print("Octobre à Décembre :")
print(oct_dec)
```

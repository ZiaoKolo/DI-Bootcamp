```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# 1. Préparation des données


# Liste des villes
villes = [
    "Abidjan", "Paris", "Londres", "New York", "Tokyo",
    "Moscou", "Caire", "Sydney", "Berlin", "Madrid"
]

# Liste des mois
mois = [
    "Jan", "Fev", "Mar", "Avr", "Mai", "Juin",
    "Juil", "Aout", "Sep", "Oct", "Nov", "Dec"
]

# Génération aléatoire des températures
temperatures = np.random.uniform(-5, 35, (10, 12))

# Création du DataFrame
df = pd.DataFrame(
    temperatures,
    index=villes,
    columns=mois
)

print("Données de température")
print(df)

# 2. Analyse des données


# Température moyenne annuelle par ville
moyennes = df.mean(axis=1)

print("Température moyenne annuelle :")
print(moyennes)

# Ville la plus chaude
ville_plus_chaude = moyennes.idxmax()

# Ville la plus froide
ville_plus_froide = moyennes.idxmin()

print("Ville la plus chaude :", ville_plus_chaude)
print("Ville la plus froide :", ville_plus_froide)


# 3. Visualisation


plt.figure(figsize=(10, 6))

for ville in villes:
    plt.plot(mois, df.loc[ville], marker='o', label=ville)

plt.title("Evolution des températures mensuelles")
plt.xlabel("Mois")
plt.ylabel("Température (°C)")
plt.legend()
plt.grid(True)

plt.show()
```

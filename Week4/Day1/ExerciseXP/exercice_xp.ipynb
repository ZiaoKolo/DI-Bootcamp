## Exercice 1

# Prédiction des Défauts de Paiement de Prêts

## Énoncé du problème

Les institutions financières accordent des prêts à de nombreux clients, mais certains ne remboursent pas leurs dettes dans les délais prévus. Cela entraîne des pertes financières importantes.

L'objectif de ce projet est de développer un modèle de Machine Learning capable de prédire si un emprunteur risque de faire défaut de paiement à partir de ses informations personnelles, financières et de son historique de crédit.

---

## Types de données nécessaires

### 1. Informations personnelles
- Âge
- Sexe
- Situation matrimoniale
- Niveau d'éducation

### 2. Informations financières
- Revenu mensuel
- Profession
- Ancienneté professionnelle
- Charges mensuelles

### 3. Informations sur le prêt
- Montant du prêt
- Durée du prêt
- Taux d'intérêt
- Type de prêt

### 4. Historique de crédit
- Score de crédit
- Nombre de crédits en cours
- Retards de paiement passés
- Taux d'endettement

### 5. Historique de remboursement
- Paiements effectués
- Paiements manqués
- Statut du prêt (remboursé ou en défaut)

---

## Sources de données

### Sources internes
- Bases de données des banques
- Historique des prêts et remboursements

### Sources externes
- Agences d'évaluation du crédit
- Organismes statistiques
- Jeux de données publics (Kaggle, UCI)

---

## Plan de collecte des données

1. Identifier les variables nécessaires.
2. Collecter les données auprès des sources internes et externes.
3. Nettoyer les données (doublons, valeurs manquantes).
4. Préparer les données pour l'analyse.
5. Diviser les données en ensembles d'entraînement et de test.
6. Vérifier la qualité et la cohérence des données.

---

## Résultat attendu

Un ensemble de données propre et structuré permettant de construire un modèle capable de prédire les défauts de paiement et d'améliorer la gestion du risque de crédit.

## Exercice 2

# Caractéristiques pertinentes pour prédire les défauts de paiement

À partir de l'ensemble de données **Prédiction de prêts**, les variables suivantes semblent les plus importantes pour prédire si un prêt sera remboursé ou non (**Loan_Status**).

| Caractéristique | Justification |
|---------------|--------------|
| **Credit_History** | Indique si le client a déjà remboursé correctement ses crédits. C'est généralement le facteur le plus déterminant du risque de défaut. |
| **ApplicantIncome** | Le revenu du demandeur influence directement sa capacité à rembourser le prêt. |
| **CoapplicantIncome** | Le revenu du co-demandeur peut renforcer la capacité de remboursement du ménage. |
| **LoanAmount** | Plus le montant emprunté est élevé, plus le risque de difficulté de remboursement peut augmenter. |
| **Loan_Amount_Term** | La durée du prêt influence le montant des mensualités et donc le risque de défaut. |
| **Dependents** | Un grand nombre de personnes à charge peut réduire la capacité financière du demandeur. |
| **Education** | Le niveau d'éducation peut être lié à la stabilité de l'emploi et aux revenus futurs. |
| **Self_Employed** | Les revenus des travailleurs indépendants peuvent être moins stables que ceux des salariés. |
| **Married** | La situation matrimoniale peut influencer la stabilité financière du foyer. |
| **Property_Area** | La zone de résidence (urbaine, semi-urbaine ou rurale) peut refléter des différences économiques et de revenus. |

## Caractéristiques moins pertinentes

- **Loan_ID** : identifiant unique du prêt, n'apporte aucune information utile pour la prédiction.
- **Gender** : généralement peu corrélé à la capacité de remboursement et peut introduire des biais.

## Conclusion

Les variables les plus importantes pour prédire les défauts de paiement sont :

1. **Credit_History**
2. **ApplicantIncome**
3. **CoapplicantIncome**
4. **LoanAmount**
5. **Loan_Amount_Term**

Ces caractéristiques décrivent directement le comportement de crédit passé du client ainsi que sa capacité financière à rembourser le prêt, ce qui en fait les meilleurs indicateurs du risque de défaut.

## Exercice 3

#  Entraînement, Évaluation et Optimisation du Modèle

## 1. Modèles adaptés à la prédiction de prêts

La prédiction de défaut de paiement est un problème de **classification binaire** (Prêt accordé/remboursé ou Défaut de paiement).

Les modèles les plus adaptés sont :

### Régression Logistique
- Simple à comprendre et à interpréter.
- Très utilisée dans le secteur bancaire.
- Bonne base de comparaison.

### Arbre de Décision
- Facile à visualiser et à expliquer.
- Capture les relations non linéaires entre les variables.

### Random Forest
- Combine plusieurs arbres de décision.
- Réduit le risque de surapprentissage.
- Généralement plus performant qu'un arbre unique.

### XGBoost
- Très performant pour les données tabulaires.
- Souvent utilisé dans les compétitions de Machine Learning.
- Offre une excellente précision.

**Choix recommandé :** Random Forest ou XGBoost, tout en utilisant la Régression Logistique comme modèle de référence.

---

## 2. Étapes d'évaluation du modèle

### Étape 1 : Séparation des données

Diviser les données en :
- 80 % pour l'entraînement
- 20 % pour le test

### Étape 2 : Entraînement du modèle

Le modèle apprend à partir des données d'entraînement afin d'identifier les caractéristiques associées aux défauts de paiement.

### Étape 3 : Prédictions

Utiliser le modèle entraîné pour prédire les résultats sur les données de test.

### Étape 4 : Évaluation des performances

Comparer les prédictions aux résultats réels à l'aide de plusieurs indicateurs.

---

## 3. Indicateurs de performance

### Accuracy (Précision globale)

Mesure le pourcentage de prédictions correctes.

\[
Accuracy = \frac{Prédictions\ correctes}{Total\ des\ prédictions}
\]

### Precision

Mesure la proportion des défauts prédits qui sont réellement des défauts.

\[
Precision = \frac{VP}{VP + FP}
\]

### Recall (Sensibilité)

Mesure la capacité du modèle à détecter les véritables défauts de paiement.

\[
Recall = \frac{VP}{VP + FN}
\]

### F1-Score

Combine Precision et Recall en une seule mesure.

\[
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
\]

### Matrice de Confusion

Permet de visualiser :
- Vrais positifs (VP)
- Faux positifs (FP)
- Vrais négatifs (VN)
- Faux négatifs (FN)

### ROC-AUC

Mesure la capacité du modèle à distinguer les emprunteurs à risque des emprunteurs fiables.

- AUC proche de 1 : excellent modèle
- AUC proche de 0,5 : modèle peu performant

---

## 4. Optimisation du modèle

Pour améliorer les performances :

- Ajuster les hyperparamètres (Grid Search ou Random Search).
- Sélectionner les variables les plus importantes.
- Traiter les valeurs manquantes.
- Équilibrer les classes si les défauts sont rares.
- Tester plusieurs modèles et comparer leurs résultats.

---

## Conclusion

Pour la prédiction des défauts de paiement, **Random Forest** et **XGBoost** sont parmi les meilleurs choix. Les performances doivent être évaluées à l'aide de l'**Accuracy**, de la **Precision**, du **Recall**, du **F1-Score** et de la **ROC-AUC**, afin d'obtenir un modèle fiable capable d'identifier efficacement les emprunteurs à risque.

## Exercice 4

# Types d'apprentissage automatique adaptés aux scénarios

## 1. Prévision des cours boursiers

### Type d'apprentissage :
**Apprentissage supervisé (Régression)**

### Justification :
Les données historiques des cours boursiers sont disponibles et les valeurs futures à prédire sont numériques. Le modèle apprend à partir des données passées pour estimer le prix futur d'une action.

**Exemples d'algorithmes :**
- Régression linéaire
- Random Forest Regressor
- XGBoost
- Réseaux de neurones (LSTM)

---

## 2. Organisation d'une bibliothèque

### Type d'apprentissage :
**Apprentissage non supervisé (Clustering)**

### Justification :
Les livres ne sont pas nécessairement étiquetés à l'avance. L'objectif est de regrouper automatiquement les ouvrages selon leurs caractéristiques communes (thème, auteur, mots-clés, contenu, etc.).

**Exemples d'algorithmes :**
- K-Means
- Clustering hiérarchique
- DBSCAN

---

## 3. Robot cherchant le chemin le plus court dans un labyrinthe

### Type d'apprentissage :
**Apprentissage par renforcement**

### Justification :
Le robot apprend en interagissant avec son environnement. Il reçoit des récompenses lorsqu'il se rapproche de la sortie et des pénalités lorsqu'il emprunte un mauvais chemin. Au fil des essais, il apprend la meilleure stratégie pour atteindre son objectif.

**Exemples d'algorithmes :**
- Q-Learning
- Deep Q-Network (DQN)
- SARSA

---

## Résumé

| Scénario | Type d'apprentissage |
|-----------|---------------------|
| Prévision des cours boursiers | Apprentissage supervisé (Régression) |
| Organisation d'une bibliothèque | Apprentissage non supervisé (Clustering) |
| Navigation d'un robot dans un labyrinthe | Apprentissage par renforcement |

## Exercice 5

# Évaluation de trois types de modèles d'apprentissage automatique

L'apprentissage automatique (*Machine Learning*) regroupe plusieurs approches permettant aux systèmes informatiques d'apprendre à partir des données. Les trois principales catégories sont l'apprentissage supervisé, l'apprentissage non supervisé et l'apprentissage par renforcement. Chaque catégorie possède des méthodes d'évaluation spécifiques adaptées à ses objectifs.

---

# 1. Modèle d'apprentissage supervisé : Classification

## Exemple

Un modèle de classification permettant de prédire si un e-mail est un spam ou non, en utilisant par exemple une régression logistique, un arbre de décision ou une forêt aléatoire.

## Stratégie d'évaluation

### a) Les métriques de performance

#### Exactitude (Accuracy)

Elle mesure la proportion des prédictions correctes.

$$
Accuracy=\frac{VP+VN}{VP+VN+FP+FN}
$$

- **VP** : vrais positifs
- **VN** : vrais négatifs
- **FP** : faux positifs
- **FN** : faux négatifs

**Avantage :**
Simple à interpréter.

**Limite :**
Peut être trompeuse lorsque les classes sont déséquilibrées.

---

#### Précision (Precision)

Elle mesure la proportion des prédictions positives correctes.

$$
Precision=\frac{VP}{VP+FP}
$$

Elle est importante lorsque les faux positifs sont coûteux.

---

#### Rappel (Recall)

Il mesure la capacité du modèle à détecter tous les cas positifs.

$$
Recall=\frac{VP}{VP+FN}
$$

Il est crucial lorsque les faux négatifs sont critiques, comme en diagnostic médical.

---

#### Score F1

Il combine précision et rappel.

$$
F1=2\times\frac{Precision\times Recall}{Precision+Recall}
$$

Il est particulièrement utile pour les jeux de données déséquilibrés.

---

### b) Méthodes d'évaluation

#### Validation croisée (*Cross-validation*)

Les données sont divisées en **k sous-ensembles**.

1. Le modèle est entraîné sur **k−1 ensembles**.
2. Il est testé sur le dernier ensemble.
3. L'opération est répétée **k fois**.

La validation croisée réduit le risque de surapprentissage et fournit une estimation plus robuste des performances.

---

#### Courbe ROC et AUC

La courbe ROC représente :

- Le taux de vrais positifs (TPR).
- Le taux de faux positifs (FPR).

L'aire sous la courbe (**AUC**) varie entre :

- **0,5** : modèle aléatoire.
- **1** : modèle parfait.

Une valeur proche de 1 indique une bonne capacité de discrimination.

---

## Difficultés et limites

- Déséquilibre des classes.
- Surapprentissage (*Overfitting*).
- Sous-apprentissage (*Underfitting*).
- Sensibilité aux données d'entraînement.
- Les métriques peuvent être contradictoires.

---

# 2. Modèle d'apprentissage non supervisé : Clustering

## Exemple

L'algorithme **K-Means** permettant de regrouper automatiquement des clients ayant des comportements similaires.

Contrairement au modèle supervisé, il n'existe généralement pas de réponse correcte connue.

## Évaluation de l'efficacité

### a) Score de silhouette

Il mesure à quel point un élément appartient à son cluster.

Le score varie entre :

- **-1** : mauvais regroupement.
- **0** : chevauchement des clusters.
- **1** : excellent regroupement.

Un score élevé indique des clusters bien séparés.

---

### b) Méthode du coude

Elle consiste à calculer la somme des distances intra-clusters pour différentes valeurs de **K**.

Le nombre optimal de clusters correspond au point où la diminution des erreurs ralentit fortement, formant un « coude ».

---

### c) Métriques de validation des clusters

#### Indice de Davies-Bouldin

Évalue :

- La compacité des clusters.
- Leur séparation.

Une faible valeur est préférable.

---

#### Indice de Calinski-Harabasz

Compare :

- La dispersion entre clusters.
- La dispersion interne.

Une valeur élevée indique une meilleure partition.

---

## Difficultés et limites

- Absence de vérité terrain.
- Choix difficile du nombre optimal de clusters.
- Sensibilité à l'initialisation.
- Résultats influencés par l'échelle des variables.
- Les métriques peuvent suggérer des nombres différents de clusters.

---

# 3. Modèle d'apprentissage par renforcement

## Exemple

Un agent apprenant à jouer à un jeu vidéo ou à piloter un robot grâce à des interactions avec son environnement.

L'agent choisit des actions et reçoit des récompenses afin d'améliorer sa stratégie.

## Mesure du succès

### a) Récompense cumulative

La performance principale est la somme des récompenses obtenues :

$$
R=\sum r_t
$$

Plus cette récompense cumulée est élevée, meilleure est la politique apprise.

---

### b) Convergence

On observe si les performances se stabilisent au fil des épisodes.

Une convergence stable indique que l'agent a appris une politique efficace.

---

### c) Équilibre entre exploration et exploitation

L'agent doit :

- Explorer de nouvelles actions.
- Exploiter les meilleures actions connues.

Un mauvais équilibre peut entraîner :

- Une convergence prématurée.
- Une exploration insuffisante.

Des stratégies comme **ε-greedy** permettent de gérer ce compromis.

---

### d) Stabilité des performances

On mesure également :

- Le nombre d'épisodes nécessaires à l'apprentissage.
- La variance des récompenses.
- La robustesse face à différents environnements.

---

## Difficultés et limites

- Apprentissage souvent très long.
- Forte dépendance à la fonction de récompense.
- Convergence non garantie.
- Grande variabilité des résultats selon les conditions initiales.
- Coût computationnel élevé.

---

# Comparaison des trois approches

| Type d'apprentissage | Exemple | Métriques principales | Principales difficultés |
|---------------------|----------|----------------------|-------------------------|
| Supervisé | Classification | Exactitude, précision, rappel, F1, ROC-AUC, validation croisée | Déséquilibre des classes, surapprentissage |
| Non supervisé | K-Means | Score de silhouette, méthode du coude, Davies-Bouldin, Calinski-Harabasz | Absence de labels, choix du nombre de clusters |
| Par renforcement | Agent intelligent | Récompense cumulative, convergence, exploration/exploitation | Temps d'apprentissage, stabilité, définition de la récompense |

---

# Conclusion

L'évaluation des modèles d'apprentissage automatique dépend fortement du type d'apprentissage utilisé. Les modèles supervisés disposent de métriques précises grâce aux données étiquetées, tandis que les modèles non supervisés s'appuient sur des mesures de cohésion et de séparation des groupes. En apprentissage par renforcement, la performance est principalement évaluée par les récompenses cumulées, la convergence et la capacité de l'agent à équilibrer exploration et exploitation. Malgré les nombreuses techniques disponibles, chaque catégorie présente des défis spécifiques, rendant indispensable l'utilisation de plusieurs critères d'évaluation afin d'obtenir une analyse fiable et complète des performances du modèle.
